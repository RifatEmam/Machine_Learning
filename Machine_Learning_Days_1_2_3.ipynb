# ***`Label Encoding`***

```
data=pd.read_csv("/content/null_removed.csv")
```

```
data.head()
```

```
data = data.drop(['Unnamed: 0'], axis=1)
```

```
data
```

```
data.dtypes
```

```
from sklearn import preprocessing
Label_encoder=preprocessing.LabelEncoder()
```

```
data['rbc_le']=Label_encoder.fit_transform(data['rbc'])
data['pc_le']=Label_encoder.fit_transform(data['pc'])
data['pcc_le']=Label_encoder.fit_transform(data['pcc'])
data['ba_le']=Label_encoder.fit_transform(data['ba'])
data['htn_le']=Label_encoder.fit_transform(data['htn'])
data['dm_le']=Label_encoder.fit_transform(data['dm'])
data['cad_le']=Label_encoder.fit_transform(data['cad'])
data['appet_le']=Label_encoder.fit_transform(data['appet'])
data['pe_le']=Label_encoder.fit_transform(data['pe'])
data['ane_le']=Label_encoder.fit_transform(data['ane'])
data['classification_le']=Label_encoder.fit_transform(data['classification'])
```

```
data
```

```
data.head()
```

```
data['classification'].unique()
```

```
data['classification_le'].unique()
```

```
data = data.drop(['rbc','pc','pcc','ba','htn','dm','cad','appet','pe','ane','classification'], axis=1, errors='ignore')
data.head()
```

```
data.to_csv("Label_Encoding_Data.csv")
```
