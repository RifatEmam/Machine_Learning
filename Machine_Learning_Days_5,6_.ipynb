# **DBSCAN**

```
import pandas as pd

df = pd.read_csv("/content/Label_Encoding_Data.csv")

from sklearn.cluster import DBSCAN
from collections import Counter

model = DBSCAN(eps=2, min_samples=1).fit(df)
outliers_df = pd.DataFrame(df)
print(Counter(model.labels_))
outlier = outliers_df[model.labels_==-1]
outlier.to_csv("/content/Label_Encoding_Data.csv")
remain = outliers_df[model.labels_!=-1]
remain.to_csv("/content/update.csv")
```
